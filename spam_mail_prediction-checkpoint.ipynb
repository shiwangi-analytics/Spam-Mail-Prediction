```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score
```

## Data collection & pre-processing

### LOADING THE DATA FROM CSV FILE TO A PANDAS DATAFRAME

```python
raw_mail_data=pd.read_csv(r"C:\Users\shiwangi\OneDrive\Desktop\ML Project@\datasets\mail_data.csv")
raw_mail_data
```

```python
### Replace the null values with a null string
```

```python
mail_data = raw_mail_data.where(pd.notnull(raw_mail_data), '')
```

```python
# print the first 5 rows of the dataframe
mail_data.head()
```

```python
#checking the no of row and columns in the dataframe
mail_data.shape
```

```python
# Label spam mail as 0; ham mail as 1;
mail_data.loc[mail_data['Category']== 'spam' , 'Category',]=0
mail_data.loc[mail_data['Category']== 'ham' , 'Category',]=1

```

```python
#separating the data as textss and label

X=mail_data['Message']

y=mail_data['Category']
```

```python
print(X)
```

```python
print(y)
```

```python
# splitting the data into traning data and test data

X_train , X_test , y_train,y_test = train_test_split(X,y , test_size=0.2 , random_state=3)
```

```python
print(X.shape)
print(X_train.shape)
print(X_test.shape)
```

```python
# transform the text data to feature vectors that can be used as input to the logistic regression

feature_extraction=TfidfVectorizer(min_df=1,stop_words='english',lowercase=True)

X_train_features=feature_extraction.fit_transform(X_train)
X_test_features=feature_extraction.transform(X_test)

# Convert y_train and y_test values as integers

y_train=y_train.astype('int')
y_test=y_test.astype('int')
```

```python
print(X_train_features)
```

## Traning the model

```python
# Logistic Regression

model=LogisticRegression()

#traning the logistic regration model with the traning data

model.fit(X_train_features,y_train)
```

```python
#Evaluation the traned model

# prediction on traning data
prediction_on_training_data=model.predict(X_train_features)
accuracy_on_traning_data=accuracy_score(y_train,prediction_on_training_data)
```

```python
print('Accuracy on traning data:',accuracy_on_traning_data)
```

```python
# prediction on test data
prediction_on_test_data=model.predict(X_test_features)
accuracy_on_test_data=accuracy_score(y_test,prediction_on_test_data)

print('Accuracy on test data:',accuracy_on_test_data)
```

## Building a predictive system

```python
input_mail = ["I've been searching for the right words to thank you for this breather"]

# Convert text to feature vectors
input_data_features = feature_extraction.transform(input_mail)

# Making prediction
prediction = model.predict(input_data_features)

print(prediction)

if prediction[0] == 1:
    print('Spam mail')
else:
    print('Ham mail')
```
